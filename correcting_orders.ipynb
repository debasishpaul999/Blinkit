```python
# loaded required libraries
import pandas as pd
```

```python
# importing required datasets
marketing = pd.read_csv('dataset/marketing.csv')
orders = pd.read_csv('dataset/orders.csv')
```

```python
orders.head()
```

```python
marketing.head()
```

```python
orders = orders.drop(columns=['campaign_id'], errors='ignore')  # dropping the false column

orders['order_date'] = pd.to_datetime(orders['order_date'],format='%Y-%m-%d',errors='coerce')  # editing the date data type
```

```python
orders.head()
```

```python
orders_base = orders.copy()

orders['key'] = 1
marketing['key'] = 1

df = orders_base.merge(marketing, on='key', how='left')
```

```python
df = df[
    (df['order_date']>=df['start_date']) &
    (df['order_date']<=df['end_date'])
]
```

```python
df.head()
```

```python
df = (
    df.sort_values(['order_id','start_date'], ascending=[True,False])
    .drop_duplicates('order_id')
)
```

```python
orders = orders_base.merge(
    df[['order_id','campaign_id']],
    on='order_id',
    how='left'
)
```

```python
orders = orders.drop(columns=['key'],errors='ignore')
marketing = marketing.drop(columns=['key'],errors='ignore')
```

```python
orders.info()
```

```python
orders.to_csv('dataset/orders.csv',index=False)
```

```python
orders = orders.merge(marketing[['campaign_id','discount_type','discount_value']], on='campaign_id',how='left')
```

```python
orders.head()
```

```python
orders['discount_amount'] = 0.0

# masks
percentage_mask = orders['discount_type'] == 'Percentage'
flat_mask = orders['discount_type'] == 'Flat'

# calculating discount amount
orders.loc[percentage_mask,'discount_amount'] = (orders.loc[percentage_mask,'total_amount'] * orders.loc[percentage_mask,'discount_value'] / 100)  # percentage discount
orders.loc[flat_mask,'discount_amount'] = orders.loc[flat_mask,'discount_value']  # flat discount

orders['discount_amount'] = orders[['total_amount','discount_amount']].min(axis=1)
```

```python
orders.info()
```

```python
orders['net_revenue'] = orders['total_amount'] + orders['delivery_fee'] - orders['discount_amount']
```

```python
orders = orders.drop(columns=['discount_type','discount_value'])
```
